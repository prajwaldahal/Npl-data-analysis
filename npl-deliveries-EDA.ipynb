## Data dictionary: `npl_deliveries.csv`

One row per delivery, 14,937 rows, wides and no-balls included. Types are what the loading cell below gives. A CSV doesn't store types, so the `category` columns are converted there after `read_csv`.

| column | dtype | definition |
|---|---|---|
| `match_id` | int64 | Cricsheet match ID; joins to `npl_matches.csv`. |
| `season` | category | Season number: 1 = 2024/25, 2 = 2025/26. |
| `innings` | int64 | 1 or 2 for the main innings. 3 and 4 are the super over in the one tied match. |
| `is_super_over` | int64 | 1 if the delivery was bowled in a super over, else 0. Filter these out for normal match stats. |
| `batting_team` | category | Code of the batting team. |
| `bowling_team` | category | Code of the bowling team. |
| `over` | int64 | Over index starting at 0, so 0 is the first over and 19 the twentieth. |
| `ball` | int64 | Position of the delivery within the over, counting wides and no-balls, so it can go above 6. |
| `actual_delivery` | float64 | Scorecard ball label, over.legal-ball: `17.5` is the fifth legal ball of the eighteenth over. A wide or no-ball has the same label as the ball bowled after it. |
| `batter` | category | Batter on strike. |
| `bowler` | str | Bowler. |
| `non_striker` | str | Batter at the non-striker's end. |
| `batter_type` | category | Batting hand of the striker: `Right-hand bat` or `Left-hand bat`. |
| `bowler_general_category` | category | `Pace` or `Spin`. |
| `bowler_type` | category | Detailed bowling style, e.g. `Right arm offbreak`, `Left arm fast medium`. |
| `runs_batter` | int64 | Runs credited to the batter off this delivery. |
| `runs_extras` | int64 | Extra runs on this delivery; the sum of `wides`, `noballs`, `byes`, `legbyes` and `penalty`. |
| `runs_total` | int64 | `runs_batter` + `runs_extras`. |
| `wides` | int64 | Runs from a wide, including any runs taken. A wide is not a legal ball. |
| `noballs` | int64 | No-ball penalty run (always 0 or 1); runs hit off a no-ball go to `runs_batter`. A no-ball is not a legal ball. |
| `byes` | int64 | Byes. |
| `legbyes` | int64 | Leg byes. |
| `penalty` | int64 | Penalty runs. 0 on every row. |
| `is_wicket` | int64 | 1 if a wicket fell on this delivery, else 0. |
| `player_out` | str | Batter dismissed; NaN when no wicket fell. |
| `dismissal_kind` | category | How they were out: `caught`, `bowled`, `lbw`, `run out`, `stumped`, `caught and bowled`, `retired out`, `hit wicket` or `retired hurt`. NaN when no wicket fell. |
| `fielder` | str | Fielder(s) involved in the dismissal, joined with `;` when there is more than one. NaN otherwise. |

Team codes: `BIK` Biratnagar Kings, `CHR` Chitwan Rhinos, `JAB` Janakpur Bolts, `KAG` Kathmandu Gorkhas, `KAY` Karnali Yaks, `LUL` Lumbini Lions, `POA` Pokhara Avengers, `SUP` Sudur Paschim Royals.

```python
import pandas as pd

category_cols = [
    "season", "batting_team", "bowling_team", "batter",
    "batter_type", "bowler_general_category", "bowler_type", "dismissal_kind",
]
deliveries = pd.read_csv("data-csv/npl_deliveries.csv")
# Convert after reading: dtype="category" in read_csv would turn season's 1 and 2 into text.
deliveries[category_cols] = deliveries[category_cols].astype("category")
```
