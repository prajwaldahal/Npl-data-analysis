## Section 3 — Team Analyst Casebook

Questions 41 to 100.

These are the questions a franchise performance analyst is actually asked: by the head coach before selection, by the bowling coach before a game, by the captain at the toss, by the recruitment team between seasons. Most answers are a table someone will act on, so the numbers need to be right before they are interesting.

### Analyst conventions

Use these definitions throughout Sections 3 and 4 unless a question says otherwise. Questions 41 to 46 ask you to build them into one enriched delivery table; everything after that builds on it.

| term | definition |
|---|---|
| ball faced (batter) | every delivery except a wide. A no-ball **is** a ball faced. |
| legal ball (bowler) | every delivery that is neither a wide nor a no-ball |
| runs conceded (bowler) | `runs_total` minus `byes` and `legbyes`. Wides and no-balls count against the bowler. |
| bowler's wicket | `is_wicket` where `dismissal_kind` is not `run out`, `retired out` or `retired hurt` |
| powerplay / middle / death | `over` 0–5 / 6–14 / 15–19. `over` is zero-indexed, so 15–19 is the 16th to 20th over. |
| strike rate | runs per 100 balls faced |
| economy | runs conceded per 6 legal balls |
| batting average | runs ÷ times out. Out means named in `player_out`, which can be the **non-striker**. |
| innings (batter) | a match-innings in which the player appeared as `batter` or `non_striker` |
| super overs | exclude them unless a question says otherwise |

Always say what minimum sample you applied. Sixty-four matches is not much data, and a ranking made from 20 balls tells you nothing. When a sample is too thin, the right answer is often "we can't tell yet".
